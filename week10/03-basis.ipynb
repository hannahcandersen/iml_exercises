**Task 1 (easy): Identifying Basis Vectors♾️**
Given the following bases vectors for $\mathbb{R}^2$:

$$
\begin{array}{l}
1. \left\{ \begin{bmatrix} 1 \\ 0 \end{bmatrix}, \begin{bmatrix} 0 \\ 1 \end{bmatrix} \right\} \\[12pt]
2. \left\{ \begin{bmatrix} 2 \\ 3 \end{bmatrix}, \begin{bmatrix} -1 \\ 2 \end{bmatrix} \right\} \\[12pt]
3. \left\{ \begin{bmatrix} 1 \\ 1 \end{bmatrix}, \begin{bmatrix} 2 \\ 2 \end{bmatrix} \right\} \\
\end{array}
$$
1. Which of these sets forms a basis for $\mathbb{R}^2 $? Justify your answer by checking linear independence and spanning the vector space.


---

```python
# To be basis vectors they need to satisfy the following conditions:

## They must be linearly independent - meaning that no vector in the set can be represented as a linear combination of the others
### If the determinant of the matrix formed by the vectors is not zero, then they are linearly independent
### det = ad - bc

## They must span the vector space - meaning that any vector in the space can be represented as a linear combination of the basis vectors
### If they are linearly independent and the number of vectors is equal to the dimension of the space, then they span the space

# Vectors 1 - They form a basis for R2: 
## Determinant: 1*1 - 0*0 = 1
## Therefore they are linearly independent
## They span the space as there are 2 vectors and the space is R2

# Vectors 2 - They form a different basis for R2:
## Determinant: 2*2 - (-1)*3 = 4 + 3 = 7
## Therefore they are linearly independent
## They span the space as there are 2 vectors and the space is R2

# Vectors 3 - They do not form a basis for R2:
## Determinant: 1*2 - 2*1 = 2 - 2 = 0
## Therefore they are not linearly independent
## Also it is obvious that v2 is a linear combination of v1 (multiply v1 with 2)
```

**Task 3 (easy): Applying a Linear Transformation in Different Bases♾️**
Let $ T: \mathbb{R}^2 \to \mathbb{R}^2 $ be a linear transformation represented by the matrix:

$$
A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}
$$

in the standard basis. Let 

$$
B = \left\{ \begin{bmatrix} 1 \\ 1 \end{bmatrix}, \begin{bmatrix} -1 \\ 2 \end{bmatrix} \right\}
$$
 
be a new basis.
1. Calculate the matrix that represents $ T $ with respect to the basis $ B $. Use the change of basis matrices to show your work.


---

```python
# Calculate matrix that represent T (tranformation using A) with respect to basis B
A = np.array([[1, 2], [3, 4]])
print("Matrix A: \n", A)
B = np.array([[1, -1], [1, 2]])
print("Matrix B: \n", B)

# change of basis matrix from E to B
TEB = B
TEB_inv = np.linalg.inv(TEB)

# Now, we can calculate the matrix Ab that represents the linear transformation T in the basis B using:
# Ab = TEB_inv * A * TEB
Ab = np.dot(np.dot(TEB_inv, A), TEB)
print("Matrix that represents T in basis B: \n", Ab)

# Convert back to standard basis
Ae = np.dot(np.dot(TEB, Ab), TEB_inv)
print("Matrix that represents T in standard basis: \n", Ae)
```

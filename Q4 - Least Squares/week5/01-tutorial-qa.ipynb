## Pen and paper exercises
A 2. order polynomial is given by 

$$f(x) = w_0 + w_1x + w_2x^2 = \sum^2_{i=0} w_ix^i.$$

Generally, an $N$'th order  polynomial is given by

$$f(x) = \sum^N_{i=0} w_ix^i,$$
where $\mathbb{w}$ is a vector of coefficients.

---

5. Generalize this algorithm to n-th order polynomials.

```python
# Write your solution here
# knowns = x, f(x)
# unknowns = w0, w1, w2

# it is linear in w because the unknowns are linear (no square or powers)
# it is non-linear in x because of x^2	

# algorithm for finding the polynomial using least squares
# define the model - y = w0 + w1*x + w2*x^2
# build the design matrix - A = [1, x, x^2]
# find P = (A^T * A)^-1 * A^T 
# solve for w: w = P * y
# to get the predicted y: y_hat = A * w

# design matrix becomes [1, x, x^2, ... x^n]
# w becomes [w0, w1, w2, ... wn]
```

#### Overview 
-   This exercise introduces matrices, their operations and relation to linear systems of equations

##### a
-   Elementwise operations
-   Matrix multiplication - a lot of dot products 
    -   If the number of columns in A equals the number of rows in B.
    -   Matrix multiplication represents linear transformations (e.g., scaling, rotation, affine mappings and projection)
-   Transpose 
-   The **inverse**: $A^{-1}A=I=AA^{-1}$
    -   For a vector to be invertible: 
        - A must be square
        - A must be full rank (all columns are linearly independent)
        - Determinant is not zero 
        - If the vectors is  minimal set of linearly independent vectors that span the space, the matrix forms a **basis** 
-   For **orthogonal matrices** the inverse is the transpose $A^TA=I=AA^T$
    - A matrix is orthogonal if 
        - All its column vectors (and rows) are orthonormal 

#### b 
-   Linear equation: $w_1x_1+w_2x_2=b+\cdots+w_nx_n = b$
-   A system of linear equations can be written in matrix form: $Ax=b$
    - Knowns in A
    - Unknowns in x
    - Output in b
- 	Find a vector x such that, after applying the transformation A, it “lands on” b.
-   To "undo" transformations: Inverse transformation: $x = A^{-1}b$
-   Only for perfectly determined sets of equation
    - Has a unique solution
	- When A is square, invertible, and spans the space, then b must be in the space ($\det(A) \neq 0$).
-   Overdetermined likely has no solution and needs an approximation like least squares
-   Underdetermined likely has many solutions and needs regularization

**Task 5 (easy): Inverse properties👩‍💻**
Use the code cell below to verify that:
1. $D^{-1}E^{-1} = (ED)^{-1}$
2. $D^{-1}E^{-1} \neq (DE)^{-1}$


---

```python
# Write your solution here
# Arrange
D_inv = np.linalg.inv(D)
E_inv = np.linalg.inv(E)
ED  = E @ D
DE = D @ E
print("D inverse\n", D_inv)
print("E inverse\n", E_inv)
print()

# Property 1
D_E_inv = D_inv @ E_inv
ED_inv = np.linalg.inv(ED)
print("Property 1")
print("D_inv @ E_inv\n", D_E_inv)
print("ED_inv\n", ED_inv)
print("D_inv @ E_inv = ED_inv:", np.allclose(D_E_inv, ED_inv, atol=1e-12))
print()

# Property 2
DE_inv = np.linalg.inv(DE)
print("Property 2")
print("D_inv @ E_inv\n", D_E_inv)
print("DE_inv\n", DE_inv)
print("D_inv @ E_inv = DE_inv:", np.allclose(D_E_inv, DE_inv, atol=1e-12))

# The properties are not equal because matrix multiplication is not commutative.
# Inverses reverse the order of multiplication.
# Multiplication order matters for matrices.
```

## Linear equations
Matrices can represent systems of linear equations

$$
Ax=b
$$
where $A$ is the coefficient matrix, $x$ vector of unknowns, and $b$ is a vector of the dependent variables.
A solution can be found using

$$
\begin{align*}
A^{-1}Ax&=A^{-1}b\\
x &= A^{-1}b.
\end{align*}
$$

## Matrix multiplication
For an $N\times D$ matrix $A$ and a $D\times K$ matrix $B$, the 
matrix multiplication (or matrix product) is a new $N\times K$ matrix $R$. Elements $R_{ij}$ of $R$ can be calculated 
using the following formula

$$
R_{ij} = \sum_{d=1}^D A_{id}B_{dj}.
$$
In other words, it is the dot product of the $i$'th row vector of $A$ and the $j$'th column vector of $B$.

---
**Task 9 (medium): Implementing matrix multiplication👩‍💻**
Implement matrix multiplication in the `matmul`
 function in the code cell below. You may use either Python lists or Numpy arrays, but the intention is to not use Numpy's built-in functions for matrix multiplication (i.e., `np.dot`
, `@`
, `np.matmul`
, etc.). You may, however, use `np.dot`
 for the purpose of computing the inner product between row and column vectors.

**Hint**
It might be helpful to calculate the correct result by hand first, to make debugging easier.


---

```python
import numpy as np

def matmul(a, b):
    N = len(a)    # Rows in A
    D = len(a[0]) # Columns in A (Rows in B)
    K = len(b[0]) # Columns in B

    # Initialize result matrix with zeros
    R = np.zeros((N, K))

    # Matrix multiplication (element-wise)
    for i in range(N):
        for j in range(K):
            for d in range(D):  # Loop through each element in the row/column
                R[i, j] += a[i][d] * b[d][j]
    
    return R

# Matrices as given in the example
ma = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

mb = [
    [5, 4, 9],
    [2, 1, 7],
    [8, 0, 1]
]

# Perform matrix multiplication
result = matmul(ma, mb)
print("Result of matrix multiplication:\n", result)
print("Result using np.dot:\n", np.dot(ma, mb))
```
